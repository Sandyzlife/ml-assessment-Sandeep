```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# load data
df = pd.read_csv('q3_retail_promotions.csv')

print(df.shape)
print(df.head())

# create new feature
# discount impact = discount * sales
df['discount_impact'] = df['discount'] * df['sales']

# basic visualization
sns.scatterplot(x='discount', y='sales', data=df)
plt.title("Discount vs Sales")
plt.show()

# scaling (feature engineering step)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.select_dtypes(include=['float64','int64']))

print("Feature engineering done")
```

A new feature called discount_impact was created to capture the effect of discounts on sales.
This helps in understanding how promotions influence revenue.
Feature scaling was applied to normalize the data for better model performance.
