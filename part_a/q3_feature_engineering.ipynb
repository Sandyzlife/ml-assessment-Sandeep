```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

# load data
df = pd.read_csv('q3_retail_promotions.csv')

print(df.shape)
print(df.head())

# create new feature (FIXED)
df['sales_impact'] = df['items_sold'] * df['competition_density']

# visualization (FIXED)
sns.scatterplot(x='competition_density', y='items_sold', data=df)
plt.title("Competition Density vs Items Sold")
plt.show()

# scaling
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.select_dtypes(include=['float64','int64']))

print("Feature engineering done")
```

A new feature called sales_impact was created using items sold and competition density.
This helps in understanding how external factors influence sales performance.
Feature scaling was applied to normalize the data.
